```python
#5 Analyze the target variable
```

```python
loan_counts = df["Personal Loan"].value_counts().sort_index()

print(loan_counts)

loan_rate = df["Personal Loan"].mean() * 100

print(f"Loan acceptance rate: {loan_rate:.2f}%")
```

```python
sns.countplot(data=df, x="Personal Loan")

plt.title("Personal Loan Acceptance")
plt.xlabel("Personal Loan (0 = No, 1 = Yes)")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```

```python
#6 Analyze income against acceptance 
```

```python
income_summary = (
    df.groupby("Personal Loan")["Income"]
      .agg(["count", "mean", "median", "min", "max"])
      .round(2)
)

display(income_summary)

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Personal Loan",
    y="Income"
)

plt.title("Income Distribution by Personal Loan Status")
plt.xlabel("Personal Loan (0 = No, 1 = Yes)")
plt.ylabel("Income")
plt.tight_layout()
plt.show()
```

```python
#7 Analyze categorical variables
```

```python
def acceptance_summary(column):
    result = (
        df.groupby(column, dropna=False)["Personal Loan"]
          .agg(Customers="size", Accepted="sum",
               Acceptance_Rate="mean")
          .reset_index()
    )

    result["Acceptance_Rate"] *= 100
    return result.round(2)
```

```python
display(acceptance_summary("Education"))
display(acceptance_summary("Family"))
display(acceptance_summary("CD Account"))
display(acceptance_summary("Securities Account"))
display(acceptance_summary("Online"))
display(acceptance_summary("CreditCard"))
```

```python
#8 Analyze credit-card spending
```

```python

display(
    df.groupby("Personal Loan")["CCAvg"]
      .agg(["count", "mean", "median"])
      .round(2)
)

sns.boxplot(
    data=df,
    x="Personal Loan",
    y="CCAvg"
)

plt.title("Credit-Card Spending by Loan Acceptance")
plt.xlabel("Personal Loan (0 = No, 1 = Yes)")
plt.ylabel("Average Monthly Spending ($000)")
plt.tight_layout()
plt.show()
```

```python
#9 Analyze relationships between numeric variables
```

```python
numeric_cols = [
    "Age", "Experience", "Income",
    "Family", "CCAvg", "Education",
    "Mortgage", "Personal Loan"
]
```

```python
corr = df[numeric_cols].corr()
```

```python
plt.figure(figsize=(10, 7))
sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    center=0
)

plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```
